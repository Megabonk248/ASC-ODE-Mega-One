```python
masses = []
for m in mss.masses:
    masses.append(
        Mesh(SphereBufferGeometry(0.2, 16, 16),
             MeshStandardMaterial(color='red'),
             position=m.pos)) 

fixes = []
for f in mss.fixes:
    fixes.append(
        Mesh(SphereBufferGeometry(0.2, 32, 16),
             MeshStandardMaterial(color='blue'),
             position=f.pos)) 

springpos = []
for s in mss.springs:
    pA = mss[s.connectors[0]].pos
    pB = mss[s.connectors[1]].pos
    springpos.append ([ pA, pB ] ) 

distanceconstraintpos = []
for s in mss.distance_constraints:
    pA = mss[s.connectors[0]].pos
    pB = mss[s.connectors[1]].pos
    distanceconstraintpos.append ([ pA, pB ] ) 

springgeo = LineSegmentsGeometry(positions=springpos)
distanceconstraintgeo = LineSegmentsGeometry(positions=distanceconstraintpos)
m2 = LineMaterial(linewidth=3, color='cyan')
m3 = LineMaterial(linewidth=3, color='red')
springs = LineSegments2(springgeo, m2)    
distanceconstraints = LineSegments2(distanceconstraintgeo, m3)

axes = AxesHelper(1)
```

```python
view_width = 600
view_height = 400

camera = PerspectiveCamera( position=[10, 6, 10], aspect=view_width/view_height)
key_light = DirectionalLight(position=[0, 10, 10])
ambient_light = AmbientLight()

scene = Scene(children=[*masses, *fixes, springs, distanceconstraints, axes, camera, key_light, ambient_light])
controller = OrbitControls(controlling=camera)
renderer = Renderer(camera=camera, scene=scene, controls=[controller],
                    width=view_width, height=view_height)

renderer
```
